```python
## Script Version 1, l'utilisateurs saisie toutes ses notes pour chaques module et le script renvoie le resultats pour son passage dans chaque UE

## definir la somme des notes et coef vides
sommeNoteUE1 = float()
sommeCoefUE1 = int()

sommeNoteUE2 = float()
sommeCoefUE2 = int()

sommeNoteUE3 = float()
sommeCoefUE3 = int()

## Calcul d'une moyenne
## moy = ((note1*coef1) + (note2*coef2) + ...)/coef1 + coef2 + ...
rLimite = 115
saeLimite = 16

## test
rLimite = 103
saeLimite = 13

## on demande à l'utilisateur chaque notes et coef des ressources
for indexR in range(101,rLimite):
    moyenneR_Actuel = float(input("Entrez votre moyenne pour la ressource R{0}".format(str(indexR))))
    coefR_ActuelUE1 = int(input("Entrez le coef UE1 pour la moyenne de la ressource actuel"))
    coefR_ActuelUE2 = int(input("Entrez le coef UE2 pour la moyenne de la ressource actuel"))
    coefR_ActuelUE3 = int(input("Entrez le coef UE3 pour la moyenne de la ressource actuel"))
    sommeNoteUE1 += moyenneR_Actuel*coefR_ActuelUE1
    sommeNoteUE2 += moyenneR_Actuel*coefR_ActuelUE2
    sommeNoteUE3 += moyenneR_Actuel*coefR_ActuelUE3

    sommeCoefUE1 += coefR_ActuelUE1
    sommeCoefUE2 += coefR_ActuelUE2
    sommeCoefUE3 += coefR_ActuelUE3

## ensuite, on demande à l'utilisateurs chaque notes et coef des SAE
for indexSAE in range(11,saeLimite):
    moyenneSAE_Actuel = float(input("Entrez votre moyenne pour la SAE{0}".format(str(indexSAE))))
    coefSAE_ActuelUE1 = int(input("Entrez le coef UE1 pour la moyenne de la SAE actuel"))
    coefSAE_ActuelUE2 = int(input("Entrez le coef UE2 pour la moyenne de la SAE actuel"))
    coefSAE_ActuelUE3 = int(input("Entrez le coef UE3 pour la moyenne de la SAE actuel"))
    sommeNoteUE1 += moyenneSAE_Actuel*coefSAE_ActuelUE1
    sommeNoteUE2 += moyenneSAE_Actuel*coefSAE_ActuelUE2
    sommeNoteUE3 += moyenneSAE_Actuel*coefSAE_ActuelUE3

    sommeCoefUE1 += coefSAE_ActuelUE1
    sommeCoefUE2 += coefSAE_ActuelUE2
    sommeCoefUE3 += coefSAE_ActuelUE3

# on crée les trois moyenne
moyenneUE1 = sommeNoteUE1/sommeCoefUE1
moyenneUE2 = sommeNoteUE2/sommeCoefUE2
moyenneUE3 = sommeNoteUE3/sommeCoefUE3

## on verifie le passage du candidat
compteurMoyenneDefaillante = 0
compteurMoyenne8_10 = 0
if moyenneUE1 < 8 : compteurMoyenneDefaillante += 1 
if moyenneUE1 < 10 and moyenneUE1 > 8 : compteurMoyenne8_10 += 1

if moyenneUE2 < 8 : compteurMoyenneDefaillante += 1 
if moyenneUE2 < 10 and moyenneUE2 > 8 : compteurMoyenne8_10 += 1
    
if moyenneUE3 < 8 : compteurMoyenneDefaillante += 1 
if moyenneUE3 < 10 and moyenneUE3 > 8 : compteurMoyenne8_10 += 1

if compteurMoyenneDefaillante > 0:
    print("Avec une moyenne de {0} pour l'UE1, {1} pour l'UE2 et {2} pour l'UE3, l'etudiant ne peut pas passer en BUT supérieur".format(moyenneUE1,moyenneUE2,moyenneUE3))

if compteurMoyenne8_10 > 0 and compteurMoyenne8_10 <= 1:
    print("Avec une moyenne de {0} pour l'UE1, {1} pour l'UE2 et {2} pour l'UE3, l'etudiant peut passer en BUT supérieur mais devra obligatoirement avoir une moyenne > 10 sur l'UE inferieure à 10".format(moyenneUE1,moyenneUE2,moyenneUE3))
if compteurMoyenne8_10 > 1:
    print("Avec une moyenne de {0} pour l'UE1, {1} pour l'UE2 et {2} pour l'UE3, l'etudiant ne peut pas passer en BUT supérieur car il a plus d'une moyenne UE inferieur à 10".format(moyenneUE1,moyenneUE2,moyenneUE3))

if compteurMoyenneDefaillante == 0 and compteurMoyenne8_10 == 0:
    print("Avec une moyenne de {0} pour l'UE1, {1} pour l'UE2 et {2} pour l'UE3, l'etudiant peut passer en BUT supérieur".format(moyenneUE1,moyenneUE2,moyenneUE3))
```

```python
## Script Version 2, les notes et les coefs sont stockés dans des dictionnaires

## creation d'un dictionnaire regroupant la liste des notes dans les ressources et dans les sae
dictNote = {
    "Ressource" : [10,11.7,18,13.9,17,13.6,19,4,15,7,16,8,11,13,12.5],
    "SAE" : [11,13,10,9,0,0]
}

## creation d'un dictionnaire regroupant la liste des coefs pour les ressources et sae de chaque UE
dictCoef = {
    "UE1" : {
        "CoefRessource" : [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0],
        "CoefSAE" : [20,20,0,0,0,7]
    },
    "UE2" : {
        "CoefRessource" : [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3],
        "CoefSAE" : [0,0,29,0,0,7]
    },
    "UE3" : {
        "CoefRessource" : [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3],
        "CoefSAE" : [0,0,0,20,20,7]
    }
}

## definir la somme des notes et coef vides
sommeNoteUE1 = float()
sommeCoefUE1 = int()

sommeNoteUE2 = float()
sommeCoefUE2 = int()

sommeNoteUE3 = float()
sommeCoefUE3 = int()

## on ajoute la somme des notes des ressources
for i in range(0,len(dictNote["Ressource"])):
    sommeNoteUE1 += dictNote["Ressource"][i]*dictCoef["UE1"]["CoefRessource"][i]
    sommeNoteUE2 += dictNote["Ressource"][i]*dictCoef["UE2"]["CoefRessource"][i]
    sommeNoteUE3 += dictNote["Ressource"][i]*dictCoef["UE3"]["CoefRessource"][i]

    sommeCoefUE1 += dictCoef["UE1"]["CoefRessource"][i]
    sommeCoefUE2 += dictCoef["UE2"]["CoefRessource"][i]
    sommeCoefUE3 += dictCoef["UE3"]["CoefRessource"][i]

## on ajoute la somme des notes des SAE
for i in range(0,len(dictNote["SAE"])):
    sommeNoteUE1 += dictNote["SAE"][i]*dictCoef["UE1"]["CoefSAE"][i]
    sommeNoteUE2 += dictNote["SAE"][i]*dictCoef["UE2"]["CoefSAE"][i]
    sommeNoteUE3 += dictNote["SAE"][i]*dictCoef["UE3"]["CoefSAE"][i]

    sommeCoefUE1 += dictCoef["UE1"]["CoefSAE"][i]
    sommeCoefUE2 += dictCoef["UE2"]["CoefSAE"][i]
    sommeCoefUE3 += dictCoef["UE3"]["CoefSAE"][i]

# on crée les trois moyenne
moyenneUE1 = round(sommeNoteUE1/sommeCoefUE1,3)
moyenneUE2 = round(sommeNoteUE2/sommeCoefUE2,3)
moyenneUE3 = round(sommeNoteUE3/sommeCoefUE3,3)

## on verifie le passage du candidat
compteurMoyenneDefaillante = 0
compteurMoyenne8_10 = 0
if moyenneUE1 < 8 : compteurMoyenneDefaillante += 1 
if moyenneUE1 < 10 and moyenneUE1 > 8 : compteurMoyenne8_10 += 1

if moyenneUE2 < 8 : compteurMoyenneDefaillante += 1 
if moyenneUE2 < 10 and moyenneUE2 > 8 : compteurMoyenne8_10 += 1
    
if moyenneUE3 < 8 : compteurMoyenneDefaillante += 1 
if moyenneUE3 < 10 and moyenneUE3 > 8 : compteurMoyenne8_10 += 1

if compteurMoyenneDefaillante > 0:
    print("Avec une moyenne de {0} pour l'UE1, {1} pour l'UE2 et {2} pour l'UE3, l'etudiant ne peut pas passer en BUT supérieur".format(moyenneUE1,moyenneUE2,moyenneUE3))

if compteurMoyenne8_10 > 0 and compteurMoyenne8_10 <= 1:
    print("Avec une moyenne de {0} pour l'UE1, {1} pour l'UE2 et {2} pour l'UE3, l'etudiant peut passer en BUT supérieur mais devra obligatoirement avoir une moyenne > 10 sur l'UE inferieure à 10".format(moyenneUE1,moyenneUE2,moyenneUE3))
if compteurMoyenne8_10 > 1:
    print("Avec une moyenne de {0} pour l'UE1, {1} pour l'UE2 et {2} pour l'UE3, l'etudiant ne peut pas passer en BUT supérieur car il a plus d'une moyenne UE inferieur à 10".format(moyenneUE1,moyenneUE2,moyenneUE3))

if compteurMoyenneDefaillante == 0 and compteurMoyenne8_10 == 0:
    print("Avec une moyenne de {0} pour l'UE1, {1} pour l'UE2 et {2} pour l'UE3, l'etudiant peut passer en BUT supérieur".format(moyenneUE1,moyenneUE2,moyenneUE3))
    
```

```python
## création d'une fonction qui permet de retourner la decision du jury sur la moyenne des trois UE

def juryDecision(listeMoy):
    compteurMoyenneDefaillante = 0
    compteurMoyenne8_10 = 0

    for i in range (3):
        if listeMoy[i] < 8 : compteurMoyenneDefaillante += 1
        if listeMoy[i] < 10 and listeMoy[i] > 8 : compteurMoyenne8_10 += 1

    if compteurMoyenneDefaillante > 0:
        return "Avec une moyenne de {0} pour l'UE1, {1} pour l'UE2 et {2} pour l'UE3, \n l'etudiant ne peut pas passer en BUT supérieur".format(moyenneUE1,moyenneUE2,moyenneUE3)

    if compteurMoyenne8_10 > 0 and compteurMoyenne8_10 <= 1:
        return "Avec une moyenne de {0} pour l'UE1, {1} pour l'UE2 et {2} pour l'UE3, \n l'etudiant peut passer en BUT supérieur mais devra obligatoirement avoir une moyenne > 10 sur l'UE inferieure à 10".format(moyenneUE1,moyenneUE2,moyenneUE3)
    if compteurMoyenne8_10 > 1:
        return "Avec une moyenne de {0} pour l'UE1, {1} pour l'UE2 et {2} pour l'UE3, \n l'etudiant ne peut pas passer en BUT supérieur car il a plus d'une moyenne UE inferieur à 10".format(moyenneUE1,moyenneUE2,moyenneUE3)

    if compteurMoyenneDefaillante == 0 and compteurMoyenne8_10 == 0:
        return "Avec une moyenne de {0} pour l'UE1, {1} pour l'UE2 et {2} pour l'UE3, \n l'etudiant peut passer en BUT supérieur".format(moyenneUE1,moyenneUE2,moyenneUE3)
```
